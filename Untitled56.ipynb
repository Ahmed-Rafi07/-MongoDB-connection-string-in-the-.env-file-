# Time-Series Anomaly Detection Pipeline
This notebook walks through loading the flights dataset, injecting controlled anomalies, and benchmarking four distinct anomaly detection approaches.

## Subtask 1: Data Preparation & Anomaly Injection

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the flights dataset
flights_raw = sns.load_dataset("flights")
# Combine year and month to create a datetime index
flights_raw['date'] = pd.to_datetime(flights_raw['year'].astype(str) + '-' + flights_raw['month'].astype(str))
flights = flights_raw.set_index('date')[['passengers']].copy()
flights.columns = ['passengers_original']

# Create a copy for manipulation
flights['passengers'] = flights['passengers_original'].astype(float)

# Inject controlled anomalies
anomalies = {
    '1952-06-01': 2.8,
    '1955-01-01': 0.2,
    '1957-08-01': 3.1,
    '1959-03-01': 0.15
}

# Create ground truth labels (0 for normal, 1 for anomaly)
true_labels = pd.Series(0, index=flights.index)

for date, factor in anomalies.items():
    target_date = pd.to_datetime(date)
    if target_date in flights.index:
        flights.loc[target_date, 'passengers'] *= factor
        true_labels.loc[target_date] = 1

flights['true_label'] = true_labels

# Plot the comparison
plt.figure(figsize=(12, 6))
plt.plot(flights.index, flights['passengers_original'], label='Original Series', color='blue', alpha=0.6)
plt.plot(flights.index, flights['passengers'], label='Manipulated Series (Injected)', color='orange', alpha=0.8)
plt.scatter(flights[flights['true_label'] == 1].index, flights[flights['true_label'] == 1]['passengers'], color='red', s=80, label='Injected Anomalies', zorder=5)
plt.title('Subtask 1: Original vs Manipulated Flights Series')
plt.xlabel('Date')
plt.ylabel('Passengers')
plt.legend()
plt.grid(True)
plt.show()
```

## Subtask 2: Method 1 — Rolling Z-Score Detection

```python
# Calculate rolling stats (window = 12, backward-aligned)
rolling_mean = flights['passengers'].rolling(window=12, center=False).mean()
rolling_std = flights['passengers'].rolling(window=12, center=False).std()

# Calculate absolute Z-score
rolling_z = (flights['passengers'] - rolling_mean) / rolling_std

# Flag anomalies using threshold of 2.5
method1_flags = (rolling_z.abs() > 2.5).astype(int)
flights['m1_zscore_flag'] = method1_flags

print(f"Total flagged points by Rolling Z-Score: {method1_flags.sum()}")
```

## Subtask 3: Method 2 — Moving Average Envelope Visualization

```python
# Construct boundaries
upper_bound = rolling_mean + (2.5 * rolling_std)
lower_bound = rolling_mean - (2.5 * rolling_std)

# Visualize
plt.figure(figsize=(12, 6))
plt.plot(flights.index, flights['passengers'], label='Manipulated Series', color='orange')
plt.plot(rolling_mean.index, rolling_mean, label='Rolling Mean (12-Month)', color='blue', linestyle='--')
plt.fill_between(flights.index, lower_bound, upper_bound, color='gray', alpha=0.2, label='Control Envelope (2.5 SD)')

# Highlight breakouts in red
breakouts = flights[method1_flags == 1]
plt.scatter(breakouts.index, breakouts['passengers'], color='red', s=100, label='Z-Score Breakouts', zorder=5)

plt.title('Subtask 3: Moving Average Envelope with Breakouts')
plt.xlabel('Date')
plt.ylabel('Passengers')
plt.legend()
plt.grid(True)
plt.show()
```

## Subtask 4: Method 3 — Robust STL Decomposition & Residual Z-Score

```python
from statsmodels.tsa.seasonal import STL

# Fit robust STL model (seasonal period of 12)
# STL requires no missing values; let's fill or verify first
series_to_decompose = flights['passengers']

stl = STL(series_to_decompose, period=12, robust=True)
res = stl.fit()

# Extract residuals
residual = res.resid

# Global Z-score on residual series
global_mean_res = residual.mean()
global_std_res = residual.std()
residual_z = (residual - global_mean_res) / global_std_res

# Flag points where absolute Z-score exceeds 3.0
method3_flags = (residual_z.abs() > 3.0).astype(int)
flights['m3_stl_flag'] = method3_flags

print("Flagged Dates by Robust STL Residual Z-Score:")
print(flights[flights['m3_stl_flag'] == 1].index.strftime('%Y-%m-%d').tolist())
```

## Subtask 5: Method 4 — Unsupervised Machine Learning (Isolation Forest)

```python
from sklearn.ensemble import IsolationForest

# Feature Engineering
features = pd.DataFrame(index=flights.index)
features['current_value'] = flights['passengers']
features['lag1'] = flights['passengers'].shift(1)
features['lag12'] = flights['passengers'].shift(12)
features['roll_mean_6'] = flights['passengers'].rolling(window=6).mean()
features['roll_std_6'] = flights['passengers'].rolling(window=6).std()

# Align ground truth labels with the feature table
features['true_label'] = flights['true_label']

# Drop initial NaN rows
features_clean = features.dropna()

X = features_clean.drop(columns=['true_label'])

# Instantiate and train Isolation Forest
isof = IsolationForest(contamination=0.05, random_state=42)
isof_preds = isof.fit_predict(X)

# Map predictions: -1 to 1 (anomaly), 1 to 0 (normal)
method4_flags_clean = np.where(isof_preds == -1, 1, 0)

# Re-align to match the original index structure (with NaNs mapped to 0)
method4_flags = pd.Series(0, index=flights.index)
method4_flags.loc[features_clean.index] = method4_flags_clean
flights['m4_isof_flag'] = method4_flags

print(f"Total flagged points by Isolation Forest: {method4_flags.sum()}")
```

## Subtask 6: Run Comparative Performance Evaluation

```python
from sklearn.metrics import precision_recall_fscore_support

# Extract aligned indices (we can score against the subset where features exist, or full index)
# Following alignment to original index:
y_true = flights['true_label']

metrics = {}
for name, col in [('Rolling Z-score', 'm1_zscore_flag'), ('STL Residual', 'm3_stl_flag'), ('Isolation Forest', 'm4_isof_flag')]:
    p, r, f, _ = precision_recall_fscore_support(y_true, flights[col], average='binary', zero_division=0)
    metrics[name] = [p, r, f]

# Create DataFrame
performance_matrix = pd.DataFrame.from_dict(
    metrics,
    orient='index',
    columns=['Precision', 'Recall', 'F1-Score']
)

# Display formatted table
display(performance_matrix.round(3))
```

## Subtask 7: The Diagnostic Reflection

Why does a simple rolling Z-score method struggle to flag sharp, contextual dips (such as our injected winter drop-offs) while the robust STL-decomposition method captures them with high precision?

Because the rolling Z-score method does not isolate the strong seasonal baseline of the dataset, leading to an inflated standard deviation during high-variance periods that masks sharp contextual dips within the wider rolling boundaries. In contrast, the robust STL-decomposition method detrends and deselects seasonal variations before analysis, cleanly isolating anomalies within the residual component with high precision.
