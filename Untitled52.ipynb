# GlobalMart Pricing & Discount Strategy Analysis

This notebook contains the full feature engineering pipeline and analysis addressing the Regional VP's key business questions.

## Subtask 1: Establish the Baseline Dataset
We will download the dataset, locate the CSV, copy/rename it to `globalmart_sales.csv`, load it, isolate the numerical columns, and verify if there are any zero-Sales records.

```python
import os
import shutil
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import kagglehub

# Download dataset
path = kagglehub.dataset_download("vivek468/superstore-dataset-final")
print("Downloaded to:", path)

# Find the CSV file
csv_files = [f for f in os.listdir(path) if f.endswith('.csv')]
if csv_files:
    src_path = os.path.join(path, csv_files[0])
    shutil.copy(src_path, 'globalmart_sales.csv')
    print("Saved dataset as globalmart_sales.csv")

# Load the baseline dataset
df = pd.read_csv('globalmart_sales.csv', encoding='windows-1252')

# Isolate numerical columns
numerical_cols = df[['Sales', 'Profit', 'Quantity', 'Discount']]
print("Numerical columns head:")
display(numerical_cols.head())

# Check for zero-Sales values
zero_sales_count = (df['Sales'] == 0).sum()
print(f"Number of records with zero Sales: {zero_sales_count}")
```

## Subtask 2: Engineer Ratio Features
We will create `profit_margin` (Profit / Sales) and `revenue_per_unit` (Sales / Quantity) ratio columns and display the first 8 rows to verify.

```python
df['profit_margin'] = df['Profit'] / df['Sales']
df['revenue_per_unit'] = df['Sales'] / df['Quantity']

display(df[['Sales', 'Profit', 'Quantity', 'profit_margin', 'revenue_per_unit']].head(8))
```

## Subtask 3: Engineer the Binary Discount Flag
We will create the binary flag `is_discounted` to show if a product was sold with a discount > 0, and output its counts.

```python
df['is_discounted'] = (df['Discount'] > 0).astype(int)
print(df['is_discounted'].value_counts())
```

## Subtask 4: Discretize Continuous Discounts
We will create the `discount_tier` categorical column using boundaries at `[-0.01, 0.0, 0.2, 0.4, 1.0]` with corresponding descriptive labels.

```python
bins = [-0.01, 0.0, 0.2, 0.4, 1.0]
labels = ['None', 'Low', 'Medium', 'High']
df['discount_tier'] = pd.cut(df['Discount'], bins=bins, labels=labels)

print(df['discount_tier'].value_counts(dropna=False))
```

## Subtask 5: Visual Exploratory Audit
Now we generate visualizations to evaluate profit margins and their relationships with Category and Discount Tiers.

```python
fig, axes = plt.subplots(1, 3, figsize=(20, 6))

# 1. Distribution of profit_margin
sns.histplot(df['profit_margin'], bins=30, kde=True, ax=axes[0], color='skyblue')
axes[0].set_title('Distribution of Profit Margin')
axes[0].set_xlabel('Profit Margin')

# 2. Box plot of profit_margin by Category
sns.boxplot(data=df, x='Category', y='profit_margin', ax=axes[1], palette='Set2')
axes[1].set_title('Profit Margin by Category')
axes[1].set_xlabel('Category')
axes[1].set_ylabel('Profit Margin')

# 3. Bar chart of mean profit margin by discount_tier with value labels
mean_profit_by_tier = df.groupby('discount_tier')['profit_margin'].mean().reset_index()
sns.barplot(data=mean_profit_by_tier, x='discount_tier', y='profit_margin', ax=axes[2], palette='viridis')
axes[2].set_title('Mean Profit Margin by Discount Tier')
axes[2].set_xlabel('Discount Tier')
axes[2].set_ylabel('Mean Profit Margin')

for p in axes[2].patches:
    axes[2].annotate(f"{p.get_height():.2%}",
                     (p.get_x() + p.get_width() / 2., p.get_height()),
                     ha='center', va='center',
                     xytext=(0, 8),
                     textcoords='offset points', fontsize=10)

plt.tight_layout()
plt.show()
```

## Subtask 6: The Regional VP's Feature Lab

### Query 1: "Which product category has the highest average revenue per unit sold — and by how much does it beat the second-highest?"

```python
rev_by_cat = df.groupby('Category')['revenue_per_unit'].mean().sort_values(ascending=False)
display(rev_by_cat)

diff = rev_by_cat.iloc[0] - rev_by_cat.iloc[1]
print(f"Difference between highest and second-highest: ${diff:.2f}")
```

Technology has the highest average revenue per unit sold at $121.36, beating the second-highest category (Furniture at $101.53) by $19.83.

### Query 2: "What proportion of orders in the South region are discounted?"

```python
south_discount_prop = df[df['Region'] == 'South']['is_discounted'].mean()
print(f"Proportion of discounted orders in South region: {south_discount_prop:.2%}")
```

In the South region, approximately 47.91% of all orders are discounted.

### Query 3: "Among orders with a profit margin below −20%, which product category appears most often?"

```python
low_margin_cats = df[df['profit_margin'] < -0.20]['Category'].value_counts()
display(low_margin_cats)
```

Office Supplies is the product category that appears most frequently (953 times) among orders with a profit margin below -20%.

### Query 4: "Does the median profit margin differ between First Class and Standard Class shipping modes?"

```python
ship_modes_comparison = df[df['Ship Mode'].isin(['First Class', 'Standard Class'])].groupby('Ship Mode')['profit_margin'].median()
display(ship_modes_comparison)
```

First Class and Standard Class shipping modes share the exact same median profit margin of 33.00%.

## Reflection

Among the findings, the most surprising result is that First Class and Standard Class shipping modes share the exact same median profit margin of 33%. One might expect faster, more expensive shipping options like First Class to command higher overall premiums and healthier profit margins, or conversely to suffer lower margins due to elevated carrier rates; however, the data shows that pricing strategies remain robustly standardized across these tiers.
