### Q1: Which product category generated the highest total revenue in each region?

```python
q1_query = """
-- Q1: Which product category generated the highest total revenue in each region?
WITH RegionalCategoryRevenue AS (
    SELECT
        dc.region,
        dp.category,
        SUM(fs.sale_amount) AS total_revenue
    FROM
        FactSales fs
    JOIN
        DimCustomer dc ON fs.customer_id = dc.customer_id
    JOIN
        DimProduct dp ON fs.product_id = dp.product_id
    GROUP BY
        dc.region, dp.category
),
RankedRegionalCategoryRevenue AS (
    SELECT
        region,
        category,
        total_revenue,
        RANK() OVER (PARTITION BY region ORDER BY total_revenue DESC) as rank_num
    FROM
        RegionalCategoryRevenue
)
SELECT
    region,
    category,
    total_revenue
FROM
    RankedRegionalCategoryRevenue
WHERE
    rank_num = 1;
"""
display(q(q1_query))
```

The output reveals that 'Electronics' is the top-performing category in the West region with a total revenue of 137,998, and also in the South region with 112,000. For the North region, 'Electronics' again leads with 108,000, while the East region's highest revenue category is also 'Electronics' at 167,000. This indicates a strong preference and sales performance for electronic products across all observed regions.

### Q2: Conditional aggregation

```python
q2_query = """
-- Q2: Calculate the total sale amount for 'Premium' customers and 'Value' customers in each region.
SELECT
    dc.region,
    SUM(CASE WHEN dc.segment = 'Premium' THEN fs.sale_amount ELSE 0 END) AS premium_customer_sales,
    SUM(CASE WHEN dc.segment = 'Value' THEN fs.sale_amount ELSE 0 END) AS value_customer_sales
FROM
    FactSales fs
JOIN
    DimCustomer dc ON fs.customer_id = dc.customer_id
GROUP BY
    dc.region;
"""
display(q(q2_query))
```

### Q3: Subquery in WHERE

```python
q3_query = """
-- Q3: Find products whose average sale amount is higher than the overall average sale amount of products in the 'Budget' price tier.
SELECT
    dp.product_name,
    dp.category,
    AVG(fs.sale_amount) AS average_sale_amount
FROM
    FactSales fs
JOIN
    DimProduct dp ON fs.product_id = dp.product_id
GROUP BY
    dp.product_name, dp.category
HAVING
    AVG(fs.sale_amount) > (
        SELECT
            AVG(fs_sub.sale_amount)
        FROM
            FactSales fs_sub
        JOIN
            DimProduct dp_sub ON fs_sub.product_id = dp_sub.product_id
        WHERE
            dp_sub.price_tier = 'Budget'
    );
"""
display(q(q3_query))
```
