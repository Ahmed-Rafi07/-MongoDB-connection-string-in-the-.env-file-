```python
import pandas as pd
import sqlite3

# --- Source 1: CSV file (patient registration desk) ---
registration_data = {
    "patient_id":     ["P001", "P002", "P003", "P004", "P005", "P002"],
    "name":           ["Meera Nair", "Arjun Shah", "Divya Rao", "Karan Mehta", "Sneha Iyer", "Arjun Shah"],
    "age":            [34, 28, 45, 52, 31, 28],
    "ward":           ["Cardiology", "Ortho", "Neurology", "Cardiology", "Ortho", "Ortho"],
    "admission_date": ["2024-03-01", "2024-03-02", "2024-03-02", "2024-03-03", "2024-03-04", "2024-03-02"]
}
reg_df = pd.DataFrame(registration_data)
reg_df.to_csv("registration.csv", index=False)
print("Registration CSV generated on disk.")

# --- Source 2: SQL database (billing department) ---
conn = sqlite3.connect("hospital_billing.db")
billing_data = {
    "patient_id":     ["P001", "P002", "P003", "P006"],
    "discharge_date": ["2024-03-05", "2024-03-06", "2024-03-07", "2024-03-08"],
    "total_bill":     ["12500", "8900", None, "15000"]
}
billing_df = pd.DataFrame(billing_data)
billing_df.to_sql("billing", conn, if_exists="replace", index=False)
conn.close()
print("Billing SQLite Database generated on disk.")
```

## ETL Pipeline Implementation

### 1. Extract: Load Data

```python
import pandas as pd
import sqlite3

# Load registration data from CSV
reg_df_extracted = pd.read_csv('registration.csv')
print("Registration data extracted:")
display(reg_df_extracted.head())

# Load billing data from SQLite database
conn = sqlite3.connect('hospital_billing.db')
billing_df_extracted = pd.read_sql_query("SELECT * FROM billing", conn)
conn.close()
print("\nBilling data extracted:")
display(billing_df_extracted.head())
```

### 2. Transform: Clean and Merge Data

```python
# Deduplicate registration data based on 'patient_id', keeping the first entry
reg_df_deduplicated = reg_df_extracted.drop_duplicates(subset='patient_id', keep='first')
print("\nRegistration data after deduplication:")
display(reg_df_deduplicated.head())

# Cast 'total_bill' to numeric, coercing errors to NaN
billing_df_extracted['total_bill'] = pd.to_numeric(billing_df_extracted['total_bill'], errors='coerce')
print("\nBilling data after casting 'total_bill' to numeric:")
display(billing_df_extracted.head())

# Merge both DataFrames on 'patient_id' using an inner join
merged_df = pd.merge(reg_df_deduplicated, billing_df_extracted, on='patient_id', how='inner')
print("\nMerged DataFrame (inner join on patient_id):")
display(merged_df.head())
```

### 3. Load: Save and Validate

```python
# Save the merged DataFrame as hospital_report.csv (excluding the index)
output_filename = "hospital_report.csv"
merged_df.to_csv(output_filename, index=False)
print(f"\nFinal report saved to '{output_filename}'.")

# Validation Assertions

# (a) Check if the row count is within the expected range
# Based on the data: reg_df unique patient_ids (P001, P002, P003, P004, P005) -> 5 rows
# billing_df patient_ids (P001, P002, P003, P006) -> 4 rows
# Inner join should result in 3 rows (P001, P002, P003)
expected_min_rows = 2
expected_max_rows = 4
assert expected_min_rows <= len(merged_df) <= expected_max_rows, \
    f"Row count ({len(merged_df)}) is not within the expected range ({expected_min_rows}-{expected_max_rows})."
print(f"Assertion (a) Passed: Row count ({len(merged_df)}) is within the expected range ({expected_min_rows}-{expected_max_rows}).")

# (b) Check if the join key (patient_id) contains zero null values
assert merged_df['patient_id'].isnull().sum() == 0, \
    f"Join key 'patient_id' contains null values (Count: {merged_df['patient_id'].isnull().sum()})."
print("Assertion (b) Passed: Join key 'patient_id' contains zero null values.")
```

## Answer 3: Markdown Analysis & Observation

The final merged report contains 3 rows, specifically for patients P001, P002, and P003. Patients P004, P005, and P006 were excluded from the `hospital_report.csv` due to the relational mechanics of an inner join. An inner join only retains rows where there is a matching value in the join key (in this case, `patient_id`) in *both* DataFrames being merged.

*   **Patient P004 and P005:** These patients are present in the `registration_df` (after deduplication) but do not have corresponding entries in the `billing_df`. Since an inner join requires a match in both sources, they were excluded.
*   **Patient P006:** This patient is present in the `billing_df` but does not have an entry in the `registration_df`. Similarly, the inner join excluded P006 for lacking a match in the registration data.

Conversely, a `left join` (or `left outer join`) would have preserved all patients from the `registration_df` (the 'left' table), including P004 and P005, and would have brought in matching billing data where available, with `NaN` values for billing columns where no match was found. Choosing the wrong join type for hospital administrative reporting carries significant operational risks. An `inner join` might inadvertently omit crucial patient records from combined reports, leading to incomplete patient histories, billing errors, or missed follow-ups for patients who only exist in one system. For instance, if P004 and P005 had open bills not yet in the billing system, an inner join would exclude them from a combined report, potentially delaying billing and revenue collection. A `left join` from registration to billing would typically be safer to ensure all registered patients are accounted for, even if their billing data is not yet available, allowing for more comprehensive administrative oversight.
