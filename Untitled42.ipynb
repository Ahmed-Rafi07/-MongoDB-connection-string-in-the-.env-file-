## Subtask 1: Structural Audit & Matrix Computation

```python
import pandas as pd

# Load the IPL dataset
df = pd.read_csv('https://raw.githubusercontent.com/12345k/IPL-Dataset/master/IPL/data.csv')

print("IPL Dataset loaded successfully.")
```

```python
# Identify and print the data types of all columns
print("Data types of all columns:")
print(df.info())
```

```python
# Filter the dataset to isolate the four key numerical variables
numerical_df = df[['win_by_runs', 'win_by_wickets', 'season', 'dl_applied']].copy()

print("Filtered numerical DataFrame head:")
display(numerical_df.head())
print("\nFiltered numerical DataFrame info:")
display(numerical_df.info())
```

```python
# Compute the full pairwise Pearson correlation matrix for these columns
correlation_matrix = numerical_df.corr(method='pearson')

# Print the matrix rounded to 2 decimal places
print("Pearson Correlation Matrix (rounded to 2 decimal places):")
display(correlation_matrix.round(2))
```

### Required Observation (Subtask 1):

Based on the correlation matrix:

*   **Strongest Statistical Correlation (off-diagonal)**: The pair `win_by_runs` and `win_by_wickets` shows the strongest negative correlation at -0.56. This suggests that as the margin of victory by runs increases, the margin of victory by wickets tends to decrease, which makes intuitive sense as a team cannot win by both a large number of runs and a large number of wickets in the same match.

*   **Weakest Statistical Correlation (off-diagonal)**: The pair `season` and `dl_applied` shows the weakest correlation at -0.01. This indicates a very negligible linear relationship between the season number and whether the Duckworth-Lewis method was applied.

*   **Perfect Self-Correlation**: As expected, all variables are perfectly correlated with themselves (correlation coefficient of 1.00), as shown on the diagonal of the matrix. This indicates that a variable is perfectly linearly related to itself.

## Subtask 2: Bivariate Visualization (Heatmap & Scatter)

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Set up the matplotlib figure size
plt.figure(figsize=(8, 6))

# Build the square-shaped seaborn heatmap
sns.heatmap(correlation_matrix.round(2), annot=True, cmap='coolwarm', fmt='.2f', linewidths=.5, center=0)

# Add a clear, professional title
plt.title('Pearson Correlation Matrix of Key Numerical Variables', fontsize=16)

# Display the plot
plt.show()
```

```python
# Identify the strongest and weakest off-diagonal correlations
# Strongest: win_by_runs vs win_by_wickets (r = -0.57)
strongest_pair_vars = ('win_by_runs', 'win_by_wickets')
strongest_r = correlation_matrix.loc[strongest_pair_vars[0], strongest_pair_vars[1]]

# Weakest: win_by_runs vs dl_applied (r = -0.01) - chosen based on absolute value closest to 0
weakest_pair_vars = ('win_by_runs', 'dl_applied')
weakest_r = correlation_matrix.loc[weakest_pair_vars[0], weakest_pair_vars[1]]

# Create scatter plot for the strongest relationship
plt.figure(figsize=(8, 6))
sns.scatterplot(data=numerical_df, x=strongest_pair_vars[0], y=strongest_pair_vars[1])
plt.title(f'Scatter Plot: {strongest_pair_vars[0]} vs {strongest_pair_vars[1]} (r = {strongest_r:.2f})', fontsize=14)
plt.xlabel(strongest_pair_vars[0].replace('_', ' ').title(), fontsize=12)
plt.ylabel(strongest_pair_vars[1].replace('_', ' ').title(), fontsize=12)
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()

# Create scatter plot for the weakest relationship
plt.figure(figsize=(8, 6))
sns.scatterplot(data=numerical_df, x=weakest_pair_vars[0], y=weakest_pair_vars[1])
plt.title(f'Scatter Plot: {weakest_pair_vars[0]} vs {weakest_pair_vars[1]} (r = {weakest_r:.2f})', fontsize=14)
plt.xlabel(weakest_pair_vars[0].replace('_', ' ').title(), fontsize=12)
plt.ylabel(weakest_pair_vars[1].replace('_', ' ').title(), fontsize=12)
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()
```

### Required Observation (Subtask 2):

**Analysis of the scatter plot for the strongest correlated pair (`win_by_runs` vs `win_by_wickets`):**

The scatter plot for `win_by_runs` versus `win_by_wickets` (r = -0.57) shows a clear and discernible pattern. The dots are not randomly scattered; instead, they generally form a **downward-sloping cloud or band**. This indicates a negative linear relationship, where as the `win_by_runs` (margin of victory by runs) increases, the `win_by_wickets` (margin of victory by wickets) tends to decrease.

Specifically, a large concentration of points is visible along both axes. When `win_by_runs` is high, `win_by_wickets` is almost always zero (indicating a win by runs), and vice-versa (when `win_by_wickets` is high, `win_by_runs` is zero, indicating a win by wickets). The scatter forms an 'L' shape, with observations clustering along the x-axis for wins by runs and along the y-axis for wins by wickets, confirming that a team can only win substantially by one metric, not both simultaneously. This pattern is consistent with the strong negative correlation found in the matrix and intuitively makes sense in the context of cricket matches.

## Final Report: IPL Matches Bivariate Correlation Audit

### Executive Summary

This audit aimed to identify meaningful, predictable relationships between key numerical performance variables in the IPL Matches dataset (`win_by_runs`, `win_by_wickets`, `season`, and `dl_applied`). We computed the Pearson correlation matrix and visualized these relationships using a heatmap and scatter plots. The analysis revealed one notable negative correlation between `win_by_runs` and `win_by_wickets`, while other pairs showed very weak or negligible linear relationships.

### Key Findings

1.  **Correlation Matrix Overview**:
    *   **Strongest Relationship**: A moderate negative correlation (r = -0.57) exists between `win_by_runs` and `win_by_wickets`. This is intuitive: a team either wins by a large margin of runs (meaning few wickets were taken by the opponent) or by a large margin of wickets (meaning the team scored runs quickly, often with runs to spare, but the opponent didn't get many wickets). It's rare for both to be high simultaneously.
    *   **Weakest Relationships**: Variables like `season` and `dl_applied`, `win_by_runs` and `dl_applied`, and `season` and `win_by_wickets` showed very weak or near-zero correlations (absolute r values close to 0.00). This suggests little to no linear relationship between these pairs.

2.  **Heatmap Visualization**:
    *   The heatmap clearly depicted the strong negative relationship between `win_by_runs` and `win_by_wickets` with a distinct cool color. The diagonal, representing self-correlation, was uniformly warm (red), as expected. All other off-diagonal cells showed colors closer to the center (white/light grey), confirming their weak linear association.

3.  **Scatter Plot Analysis**:
    *   **Strongest Pair (`win_by_runs` vs `win_by_wickets`, r = -0.57)**: The scatter plot for this pair showed a clear 'L' shaped pattern, with points clustering along the axes rather than forming a single linear slope. This reflects the reality of cricket outcomes: wins are typically either by a significant margin of runs (zero or very few wickets taken by the opponent) or by a significant margin of wickets (winning target reached with many wickets in hand). This pattern confirms the negative correlation but also illustrates its specific contextual meaning.
    *   **Weakest Pair (`win_by_runs` vs `dl_applied`, r = -0.01)**: The scatter plot for this pair showed a random cloud of points with no discernible pattern, consistent with the near-zero correlation. This indicates that the margin of victory by runs has no linear predictive power over whether the Duckworth-Lewis method was applied in a match.

### Warning: Misleading Patterns and Spurious Correlations

Client, it is crucial to understand the limitations of bivariate correlation analysis, especially when identifying 'predictable relationships':

1.  **Correlation Does Not Imply Causation**: Even a strong correlation, like the one between `win_by_runs` and `win_by_wickets`, does not mean one variable *causes* the other. In this case, both are outcomes of the same underlying event (a cricket match) and represent different facets of victory. They are mutually exclusive by definition in their extreme forms.

2.  **Linearity Assumption**: Pearson correlation measures the strength of *linear* relationships. Complex non-linear relationships might exist but would not be captured by this metric. For example, some threshold effect might be at play that a linear correlation would miss.

3.  **Third Variables/Confounders**: Apparent correlations might be influenced by unmeasured 'third variables' or confounding factors. For instance, the weather might influence `dl_applied` but also indirectly affect scoring rates, creating an indirect link not captured by direct bivariate analysis.

4.  **Sampling Variability**: Correlations observed in this dataset are based on a specific sample of IPL matches. While the dataset is substantial, these coefficients are estimates and might vary slightly in other samples or future seasons.

5.  **Context is Key**: The strong negative correlation between `win_by_runs` and `win_by_wickets` is not a 'predictive relationship' in the traditional sense, but rather a descriptive outcome of cricket rules. It doesn't tell us how to *achieve* a win, but rather *how* a win was achieved. Focusing solely on this high correlation to predict future match outcomes without considering other game factors would be misleading.
