```python
import pandas as pd
import numpy as np

url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"
df_raw = pd.read_csv(url)

print(f"Raw dataset: {df_raw.shape[0]} rows, {df_raw.shape[1]} columns")
print("Null counts -the five problems Module 2 addressed:")
print(df_raw.isna().sum()[df_raw.isna().sum() > 0])
print(f"Fare range: £{df_raw['Fare'].min():.2f} to £{df_raw['Fare'].max():.2f}")
print(f"Age range:  {df_raw['Age'].min():.1f} to {df_raw['Age'].max():.1f} years")
```

```python
Q1 = df_raw['Fare'].quantile(0.25)
Q3 = df_raw['Fare'].quantile(0.75)
IQR = Q3 - Q1
lower_fence = Q1 - 1.5 * IQR
upper_fence = Q3 + 1.5 * IQR

# Step 3: Flag outliers
fare_outliers = df_raw[(df_raw['Fare'] < lower_fence) | (df_raw['Fare'] > upper_fence)]

print(f"Q1 (25th percentile): £{Q1:.2f}")
print(f"Q3 (75th percentile): £{Q3:.2f}")
print(f"IQR (Q3 minus Q1):    £{IQR:.2f}")
print()
print(f"Lower fence (Q1 - 1.5 * IQR): £{lower_fence:.2f}")
print(f"Upper fence (Q3 + 1.5 * IQR): £{upper_fence:.2f}")
print()
print(f"Passengers flagged as outliers: {len(fare_outliers)}")
print(f"Highest fare flagged: £{fare_outliers['Fare'].max():.2f}")
print(f"Lowest fare flagged:  £{fare_outliers['Fare'].min():.2f}")
print()
print("Pclass breakdown of flagged passengers:")
print(fare_outliers['Pclass'].value_counts().sort_index())
```

```python
df_test = df_raw.copy()  # always work on a copy
df_test['Fare'] = df_test['Fare'].clip(lower=0, upper=upper_fence)

print("=== Effect of capping ===")
print(f"Before -max: £{df_raw['Fare'].max():.2f}  mean: £{df_raw['Fare'].mean():.2f}")
print(f"After  -max: £{df_test['Fare'].max():.2f}  mean: £{df_test['Fare'].mean():.2f}")
print()
print(f"Rows where Fare changed: {(df_raw['Fare'] != df_test['Fare']).sum()}")
print(f"Rows where Fare unchanged: {(df_raw['Fare'] == df_test['Fare']).sum()}")
print()
print("Sample of capped values (showing original and capped side by side):")
changed = df_raw[df_raw['Fare'] != df_test['Fare']][['Fare', 'Pclass']].copy()
changed['Fare_capped'] = df_test.loc[changed.index, 'Fare']
print(changed.head(5).to_string())
```

```python
def clean_titanic(df, verbose=False):
    df_cleaned = df.copy()

    # Example: Impute missing Age values with the median
    if 'Age' in df_cleaned.columns and df_cleaned['Age'].isnull().sum() > 0:
        median_age = df_cleaned['Age'].median()
        df_cleaned['Age'].fillna(median_age, inplace=True)
        if verbose:
            print(f"Filled {df['Age'].isnull().sum()} missing Age values with median ({median_age:.1f})")

    # Example: Impute missing Embarked values with the mode
    if 'Embarked' in df_cleaned.columns and df_cleaned['Embarked'].isnull().sum() > 0:
        mode_embarked = df_cleaned['Embarked'].mode()[0]
        df_cleaned['Embarked'].fillna(mode_embarked, inplace=True)
        if verbose:
            print(f"Filled {df['Embarked'].isnull().sum()} missing Embarked values with mode ({mode_embarked})")

    # Example: Drop Cabin column due to high number of missing values
    if 'Cabin' in df_cleaned.columns and df_cleaned['Cabin'].isnull().sum() > 0.7 * len(df_cleaned):
        df_cleaned.drop('Cabin', axis=1, inplace=True)
        if verbose:
            print("Dropped 'Cabin' column due to too many missing values")

    # Example: Apply Fare capping using previously calculated upper_fence
    if 'Fare' in df_cleaned.columns:
        # Ensure upper_fence is defined. If not, calculate it here or handle gracefully.
        # For this example, assuming upper_fence is available from previous execution
        global upper_fence # Access the global variable defined in a prior cell
        if 'upper_fence' in globals():
            df_cleaned['Fare'] = df_cleaned['Fare'].clip(lower=0, upper=upper_fence)
            if verbose:
                print(f"Capped 'Fare' column to max £{upper_fence:.2f}")
        else:
            if verbose:
                print("Warning: 'upper_fence' not defined for Fare capping. Skipping capping.")


    return df_cleaned

df_clean = clean_titanic(df_raw, verbose=True)

print(" " + "="*50)
print("PRE-PIPELINE AUDIT (df_raw)")
print("="*50)
print(f"Shape:          {df_raw.shape}")
print(f"Total nulls:    {df_raw.isna().sum().sum()}")
print(f"Max Fare:       £{df_raw['Fare'].max():.2f}")
print(f"Columns:        {list(df_raw.columns)}")

print(" " + "="*50)
print("POST-PIPELINE AUDIT (df_clean)")
print("="*50)
print(f"Shape:          {df_clean.shape}")
print(f"Total nulls:    {df_clean.isna().sum().sum()}")
print(f"Max Fare:       £{df_clean['Fare'].max():.2f}")
new_cols = [c for c in df_clean.columns if c not in df_raw.columns]
print(f"New columns:    {new_cols}")

print("" + "="*50)
print("IDEMPOTENCY TEST")
print("="*50)
# Run the pipeline again on the already-cleaned output
# If the pipeline is idempotent, results must match the first run exactly
df_twice = clean_titanic(df_clean)
print(f"Run once:  shape={df_clean.shape}, nulls={df_clean.isna().sum().sum()}")
print(f"Run twice: shape={df_twice.shape}, nulls={df_twice.isna().sum().sum()}")
print(f"Idempotency: {'PASS -identical output' if df_clean.shape == df_twice.shape else 'FAIL'}")
```
