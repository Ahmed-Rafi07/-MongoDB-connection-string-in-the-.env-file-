### Q1 — Roll-Up

```python
# Q1: Roll-up with ROLLUP
q("""
SELECT
  c.region,
  p.category,
  SUM(fs.sale_amount) AS total_sales
FROM FactSales fs
JOIN DimCustomer c ON fs.customer_id = c.customer_id
JOIN DimProduct p ON fs.product_id = p.product_id
GROUP BY ROLLUP (c.region, p.category)
ORDER BY c.region ASC, p.category ASC;
""")
```

### Q2 — Drill-Down

```python
# Q2: Drill-Down
q("""
SELECT
  c.region,
  d.quarter,
  SUM(fs.sale_amount) AS total_sales
FROM FactSales fs
JOIN DimCustomer c ON fs.customer_id = c.customer_id
JOIN DimDate d ON fs.date_id = d.date_id
WHERE c.region = 'West'
GROUP BY c.region, d.quarter
ORDER BY d.quarter ASC;
""")
```

### Q3 — Slice

```python
# Q3: Slice
q("""
SELECT
  p.category,
  SUM(fs.sale_amount) AS total_sales
FROM FactSales fs
JOIN DimProduct p ON fs.product_id = p.product_id
JOIN DimCustomer c ON fs.customer_id = c.customer_id
WHERE c.region = 'South'
GROUP BY p.category
ORDER BY total_sales DESC;
""")
```

### Q4 — Dice

```python
# Q4: Dice
q("""
SELECT
  c.region,
  d.quarter,
  SUM(fs.sale_amount) AS total_sales
FROM FactSales fs
JOIN DimCustomer c ON fs.customer_id = c.customer_id
JOIN DimDate d ON fs.date_id = d.date_id
WHERE c.region IN ('North', 'East') AND d.quarter IN ('Q1', 'Q2')
GROUP BY c.region, d.quarter
ORDER BY c.region ASC, d.quarter ASC;
""")
```

This diced view answers the business question: "What were the total sales in the North and East regions during the first two quarters (Q1 and Q2)?" The results show that the North region had sales of 1200 in Q1 and 899 in Q2, while the East region had sales of 68000 in Q1 and 4200 in Q2, highlighting a significantly higher Q1 performance for the East region compared to the North.
