# Patient Feedback Text EDA Report

## Subtask 1: Build the Patient Feedback Corpus

```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import re
import string
from collections import Counter

# Set random seed
np.random.seed(0)

# Load dataset
tips = sns.load_dataset('tips')

# Handle missing values by filling with medians (safety measure)
tips['total_bill'] = tips['total_bill'].fillna(tips['total_bill'].median())
tips['size'] = tips['size'].fillna(tips['size'].median())

def make_feedback(row):
    templates = [
        "Patient visited on a {day}. Party size was {size}. Total bill: {bill:.2f}.",
        "A {time} appointment on {day}. Bill of {bill:.2f} for {size} people.",
        "Visited during {time} on {day}. Party of {size}, billed {bill:.2f}."
    ]
    template = np.random.choice(templates)
    return template.format(day=row['day'], size=row['size'], bill=row['total_bill'], time=row['time'])

# Construct feedback column
tips['feedback'] = tips.apply(make_feedback, axis=1)

# Keep selected columns
df = tips[['feedback', 'sex', 'day', 'time', 'size']].copy()

# Print shape and preview
print("Dataset Shape:", df.shape)
display(df.head(4))
```

## Subtask 2: Text Preprocessing

```python
# Define Stopwords set
STOPWORDS = {'the', 'a', 'an', 'is', 'was', 'in', 'of', 'at', 'to', 'for', 'on', 'by', 'and', 'they', 'this', 'it', 'with', 'or', 'but', 'not'}

def preprocess(text):
    # 1. Lowercase
    text_lower = text.lower()
    # 2. Remove punctuation
    translator = str.maketrans('', '', string.punctuation)
    text_clean = text_lower.translate(translator)
    # 3. Split and filter stopwords
    tokens = [word for word in text_clean.split() if word not in STOPWORDS]
    return tokens

# Apply preprocess
df['tokens'] = df['feedback'].apply(preprocess)

# Verification
print("Original Feedback:\n", df['feedback'].iloc[0])
print("\nPreprocessed Tokens:\n", df['tokens'].iloc[0])
```

## Subtask 3: Basic Text Statistics & Quality Check

```python
# Part A: Compute statistics
df['char_count'] = df['feedback'].apply(len)
df['word_count'] = df['feedback'].apply(lambda x: len(x.split()))
df['sentence_count'] = df['feedback'].apply(lambda x: len(re.findall(r'[.!?]', x)))
df['avg_word_length'] = df['tokens'].apply(lambda t: np.mean([len(w) for w in t]) if len(t) > 0 else 0)

print("Dataframe Head with Statistics:")
display(df[['feedback', 'char_count', 'word_count', 'sentence_count', 'avg_word_length']].head())

print("\nDescriptive Statistics:")
display(df[['char_count', 'word_count', 'sentence_count']].describe())
```

```python
# Part B: Distributions
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

sns.histplot(df['char_count'], bins=15, ax=axes[0], color='skyblue', kde=True)
axes[0].set_title('Distribution of Character Count')
axes[0].set_xlabel('Character Count')
axes[0].set_ylabel('Frequency')

sns.histplot(df['word_count'], bins=15, ax=axes[1], color='salmon', kde=True)
axes[1].set_title('Distribution of Word Count')
axes[1].set_xlabel('Word Count')
axes[1].set_ylabel('Frequency')

sns.histplot(df['avg_word_length'], bins=15, ax=axes[2], color='lightgreen', kde=True)
axes[2].set_title('Distribution of Avg Word Length')
axes[2].set_xlabel('Avg Word Length')
axes[2].set_ylabel('Frequency')

plt.tight_layout()
plt.show()
```

```python
# Part C: Quality Checks
print("=== Quality Checks ===")

# 1. Count empty/null
empty_count = df['feedback'].isnull().sum() + (df['feedback'].str.strip() == '').sum()
print(f"Empty or Null Feedback Entries: {empty_count}")

# 2. Length outliers (99th and 1st percentile)
p99 = df['word_count'].quantile(0.99)
p01 = df['word_count'].quantile(0.01)
outliers_high = df[df['word_count'] > p99]
outliers_low = df[df['word_count'] < p01]
print(f"Reviews above 99th percentile (> {p99} words): {len(outliers_high)}")
print(f"Reviews below 1st percentile (< {p01} words): {len(outliers_low)}")

# 3. Non-ASCII characters
non_ascii_count = df['feedback'].apply(lambda x: bool(re.search(r'[^\x00-\x7F]', x))).sum()
print(f"Reviews containing non-ASCII characters: {non_ascii_count}")
```

## Subtask 4: Term Frequency Analysis

```python
# Flatten token list
all_tokens = [token for sublist in df['tokens'] for token in sublist]

# Count frequencies
token_counts = Counter(all_tokens)
common_tokens = token_counts.most_common(20)

words = [item[0] for item in common_tokens]
counts = [item[1] for item in common_tokens]

# Plot
plt.figure(figsize=(12, 6))
bars = plt.bar(words, counts, color='purple')
plt.title('Top 20 Most Frequent Words', fontsize=14)
plt.xlabel('Words', fontsize=12)
plt.ylabel('Frequency', fontsize=12)
plt.xticks(rotation=45, ha='right')

# Add count labels on top of bars
for bar in bars:
    yval = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2.0, yval + 1, str(yval), ha='center', va='bottom', fontsize=9)

plt.tight_layout()
plt.show()

print(f"Total Tokens (Corpus Size): {len(all_tokens)}")
print(f"Total Unique Tokens (Vocabulary Size): {len(token_counts)}")
```

## Subtask 5: N-Gram Analysis — Bigrams and Trigrams

```python
def get_ngrams(token_list, n):
    return [tuple(token_list[i:i+n]) for i in range(len(token_list) - n + 1)]

# Generate all bigrams and trigrams
all_bigrams = [bg for tokens in df['tokens'] for bg in get_ngrams(tokens, 2)]
all_trigrams = [tg for tokens in df['tokens'] for tg in get_ngrams(tokens, 3)]

bigram_counts = Counter(all_bigrams).most_common(15)
trigram_counts = Counter(all_trigrams).most_common(15)

# Format strings for visualization
bg_labels = [" ".join(item[0]) for item in bigram_counts][::-1]
bg_values = [item[1] for item in bigram_counts][::-1]

tg_labels = [" ".join(item[0]) for item in trigram_counts][::-1]
tg_values = [item[1] for item in trigram_counts][::-1]

# Horizontal bar charts side by side
fig, axes = plt.subplots(1, 2, figsize=(16, 7))

axes[0].barh(bg_labels, bg_values, color='teal')
axes[0].set_title('Top 15 Bigrams')
axes[0].set_xlabel('Frequency')

axes[1].barh(tg_labels, tg_values, color='coral')
axes[1].set_title('Top 15 Trigrams')
axes[1].set_xlabel('Frequency')

plt.tight_layout()
plt.show()
```

## Subtask 6: Zipf's Law Verification

```python
sorted_freq = sorted(list(token_counts.values()), reverse=True)
ranks = np.arange(1, len(sorted_freq) + 1)

# Plot
fig, axes = plt.subplots(1, 2, figsize=(14, 6))

# Linear plot
axes[0].plot(ranks, sorted_freq, marker='o', color='b', linestyle='-')
axes[0].set_title('Rank vs Frequency (Linear Scale)')
axes[0].set_xlabel('Rank')
axes[0].set_ylabel('Frequency')

# Log-log plot
axes[1].loglog(ranks, sorted_freq, marker='o', color='r', linestyle='None')
axes[1].set_title("Zipf's Law (Log-Log Scale)")
axes[1].set_xlabel('Log Rank')
axes[1].set_ylabel('Log Frequency')

plt.tight_layout()
plt.show()

# Hapax legomena count
hapax_legomena = sum(1 for f in sorted_freq if f == 1)
print(f"Total Vocabulary Size: {len(sorted_freq)}")
print(f"Hapax Legomena (words appearing exactly once): {hapax_legomena}")
```

## Subtask 7: Word Frequency Heatmap by Group

```python
# Determine overall top-10 words
top_10_words = [item[0] for item in token_counts.most_common(10)]
unique_groups = df['time'].unique()

heatmap_data = {}

for group in unique_groups:
    group_tokens = [tok for tokens in df[df['time'] == group]['tokens'] for tok in tokens]
    total_group_tokens = len(group_tokens)
    group_counter = Counter(group_tokens)

    group_rates = []
    for word in top_10_words:
        if total_group_tokens > 0:
            # Occurrence rate per 1,000 tokens
            rate = (group_counter[word] / total_group_tokens) * 1000
        else:
            rate = 0.0
        group_rates.append(rate)

    heatmap_data[group] = group_rates

# Build DataFrame for visualization
heatmap_df = pd.DataFrame(heatmap_data, index=top_10_words).T

plt.figure(figsize=(10, 5))
sns.heatmap(heatmap_df, annot=True, fmt='.1f', cmap='Blues', linewidths=0.5)
plt.title('Normalized Word Frequency (per 1,000 tokens) by Meal Time')
plt.xlabel('Top 10 Words')
plt.ylabel('Meal Time')
plt.tight_layout()
plt.show()
```

## Subtask 8: Analysis & Insights

1. **Preprocessing Evaluation**:
   - Cleaning punctuation and converting words to lowercase effectively standardizes variables like `day` and `time` regardless of where they appeared in the template sentences.
   - Removing English helper stopwords ensures that key metrics, like term frequencies, capture semantic vocabulary trends (e.g. "visited", "billed", "patient") rather than grammar structure.

2. **Corpus Characteristics & Constraints**:
   - This synthetic corpus shows uniform distributions across metric counts because templates restrict the linguistic variety. Real patient feedback tends to have heavy right-skewed length distributions.
   - In real-world data, spelling errors, colloquial jargon, and non-ASCII (such as emojis or accented characters) would be present and would require custom normalization strategies.

3. **Domain Adaptation & Clinical Applications**:
   - In a clinical tech setting, analyzing bigrams/trigrams directly alerts managers to systematic operational bottlenecks, such as "long wait time", "poor communication", or "billing errors".
   - Stratifying these phrases across groups (e.g., department or appointment time) enables the organization to design targeted training and improvements directly corresponding to patient complaints.
