## Subtask 1: Establish the Baseline Pipeline

In this section, we define the base cleaning function, retrieve the Titanic dataset, and extend the pipeline to address hidden missing values (Zero Fares) using group-conditional median imputation.

```python
import pandas as pd
import numpy as np

def clean_titanic(url):
    # Stage 1: Load Data
    df = pd.read_csv(url)

    # Stage 2: Deduplication
    df = df.drop_duplicates()

    # Stage 3: Type Coercion
    df['Survived'] = df['Survived'].astype('category')
    df['Pclass'] = df['Pclass'].astype('category')

    # Stage 4: Structural Tweaks
    df['has_cabin'] = df['Cabin'].notna().astype(int)
    df = df.drop(columns=['Cabin', 'Ticket'])

    # Stage 5: String Normalization
    df['Name'] = df['Name'].str.strip().str.title()
    df['Sex'] = df['Sex'].str.strip().str.lower()
    df['Embarked'] = df['Embarked'].str.strip().str.upper()

    # Stage 6: Missing Value Imputation
    # Age imputation based on Pclass median
    df['Age'] = df.groupby('Pclass')['Age'].transform(lambda x: x.fillna(x.median()))

    # Embarked mode imputation
    embarked_mode = df['Embarked'].mode()[0]
    df['Embarked'] = df['Embarked'].fillna(embarked_mode)

    # Stage 1b: Hidden Missing Values (Fare Column)
    # Replace 0 with NaN
    df['Fare'] = df['Fare'].replace(0, np.nan)

    # Group-conditional median based on Pclass to fill Fare gaps
    df['Fare'] = df.groupby('Pclass')['Fare'].transform(lambda x: x.fillna(x.median()))

    return df

titanic_url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"
df_clean = clean_titanic(titanic_url)
print("Pipeline completed successfully!")
```

### Observation - Fare Treatment Justification

**Justification of Treatment:**
Passengers with a Fare of 0 represent instances where transaction amounts were either omitted, errored out during data entry, or represent complimentary tickets. Leaving them as 0 falsely drags down statistical summaries like average ticket cost.

**Classification of "Zero Fare" Mechanism:**
This is classified as **MAR (Missing at Random)**. The missingness (Zero Fares) is likely correlated with other observed variables such as passenger class, ticket type, or booking status (e.g., ship crew, company employees, or promotional bookings) rather than the missing fare value itself.

**Imputation Method & Bias Prevention:**
Using group-conditional median based on `Pclass` preserves the characteristic scale of economic segments on the ship. Simply using a global mean/median would introduce analytical bias by over-inflating Third Class tickets or under-valuing First Class cabins. Imputing conditional medians minimizes outlier impact and maintains variance structures within the groups.

## Subtask 2: Fuzzy Matching Audit on Surnames

Here, we extract passenger surnames, isolate the top 5 most frequent last names, and utilize a sequence matching algorithm to inspect the dataset for potential spelling duplicates.

```python
from difflib import SequenceMatcher

def similarity_ratio(a, b):
    return SequenceMatcher(None, a, b).ratio()

# Extract Surnames
df_clean['Surname'] = df_clean['Name'].str.split(',').str[0]

# Identify Common Surnames
top_5_surnames = df_clean['Surname'].value_counts().head(5).index.tolist()
print(f"Top 5 most frequent surnames: {top_5_surnames}\n")

# Run Fuzzy Check
print("Fuzzy Matching Comparison Matrix:")
for i in range(len(top_5_surnames)):
    for j in range(i + 1, len(top_5_surnames)):
        name1 = top_5_surnames[i]
        name2 = top_5_surnames[j]
        score = similarity_ratio(name1, name2)
        print(f"Comparing {name1} and {name2}: Similarity Score = {score:.4f}")
```

### Observation - Fuzzy Matching Findings

**Findings:**
Comparing the top 5 surnames (e.g., Andersson, Sage, Goodwin, Skoog, Carter) yields low similarity scores, indicating they are genuinely distinct. No two names exceed the fuzzy threshold of 0.85.

**Why Fuzzy Matching is Safer:**
Simple equality checks (`==`) only capture exact spelling. Typographical errors (like 'Anderson' vs 'Andersson') have a high symmetric difference in raw characters. Fuzzy matching allows analysts to catch phonetically or visually similar string anomalies without requiring identical exact values, proving to be a much safer diagnostic tool for dirty categorical inputs.
