# GlobalMart Customer Segmentation Analysis
This notebook contains the complete K-Means clustering pipeline to segment customer orders, profile segments, and deliver actionable marketing recommendations.

## Subtask 1: Setup & Feature Standardization

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA

# 1. Load the dataset (Make sure globalmart_sales.csv is uploaded)
try:
    df = pd.read_csv('globalmart_sales.csv')
    print("Dataset successfully loaded. Shape:", df.shape)
except FileNotFoundError:
    print("Error: Please upload 'globalmart_sales.csv' to the Colab environment.")
    # Generate dummy data for demonstration if file not uploaded yet
    np.random.seed(42)
    dummy_data = {
        'Sales': np.random.exponential(scale=200, size=1000) + 10,
        'Profit': np.random.normal(loc=20, scale=50, size=1000),
        'Quantity': np.random.randint(1, 15, size=1000),
        'Discount': np.random.choice([0.0, 0.1, 0.2, 0.5, 0.7], size=1000, p=[0.5, 0.2, 0.15, 0.1, 0.05])
    }
    df = pd.DataFrame(dummy_data)
    print("Created mock dataset with shape:", df.shape)

# 2. Recreate the four derived features
df['profit_margin'] = df['Profit'] / df['Sales']
df['revenue_per_unit'] = df['Sales'] / df['Quantity']
df['is_discounted'] = df['Discount'] > 0

def get_discount_tier(d):
    if d == 0: return 'None'
    elif d <= 0.2: return 'Low'
    elif d <= 0.5: return 'Medium'
    else: return 'High'

df['discount_tier'] = df['Discount'].apply(get_discount_tier)

# 3. Standardize the four original numerical columns
features_to_scale = ['Sales', 'Profit', 'Quantity', 'Discount']
scaler = StandardScaler()
scaled_features = scaler.fit_transform(df[features_to_scale])

# Save scaled features back into a new DataFrame representation for K-Means
df_scaled = pd.DataFrame(scaled_features, columns=[f'scaled_{col}' for col in features_to_scale])
display(df.head())
display(df_scaled.head())
```

## Subtask 2: Generate the Elbow Plot

```python
inertias = []
K_range = range(1, 11)

for k in K_range:
    kmeans = KMeans(n_clusters=k, random_state=42, n_init='auto')
    kmeans.fit(df_scaled)
    inertias.append(kmeans.inertia_)

plt.figure(figsize=(8, 5))
plt.plot(K_range, inertias, marker='o', linestyle='-', color='b')
plt.title('Elbow Method for Optimal K')
plt.xlabel('Number of Clusters (K)')
plt.ylabel('Inertia')
plt.grid(True)
plt.show()
```

## Subtask 3: Fit K-Means & Label the Dataset

```python
kmeans_opt = KMeans(n_clusters=4, random_state=42, n_init='auto')
df['Cluster'] = kmeans_opt.fit_predict(df_scaled)
print("Cluster counts:\n", df['Cluster'].value_counts())
print("Labels assigned successfully.")
```

## Subtask 4: Compute the Segment Profile Table

```python
profile_table = df.groupby('Cluster').agg(
    order_count=('Cluster', 'count'),
    mean_sales=('Sales', 'mean'),
    mean_profit=('Profit', 'mean'),
    mean_quantity=('Quantity', 'mean'),
    mean_discount=('Discount', 'mean'),
    mean_profit_margin=('profit_margin', 'mean'),
    mean_revenue_per_unit=('revenue_per_unit', 'mean'),
    percent_discounted=('is_discounted', lambda x: (x.sum() / len(x)) * 100)
).reset_index()

pd.set_option('display.max_columns', None)
display(profile_table)
```

## Subtask 5: Visualize Individual Feature Distributions

```python
fig, axes = plt.subplots(2, 2, figsize=(14, 10))
features = ['Sales', 'Profit', 'Quantity', 'Discount']

for idx, col in enumerate(features):
    ax = axes[idx // 2, idx % 2]
    sns.boxplot(x='Cluster', y=col, data=df, ax=ax, palette='Set2', hue='Cluster', legend=False)
    ax.set_title(f'Distribution of {col} by Cluster')
    ax.grid(True, linestyle='--', alpha=0.5)

plt.tight_layout()
plt.show()
```

## Subtask 6: Visualize Profit Margins & Clusters in PCA Space

```python
# Perform PCA to reduce scaled features to 2 components
pca = PCA(n_components=2, random_state=42)
pca_components = pca.fit_transform(df_scaled)
df['PCA1'] = pca_components[:, 0]
df['PCA2'] = pca_components[:, 1]

fig, axes = plt.subplots(1, 2, figsize=(16, 6))

# Left Panel: Box Plot of profit_margin by cluster with a horizontal break-even line
sns.boxplot(x='Cluster', y='profit_margin', data=df, ax=axes[0], palette='Set2', hue='Cluster', legend=False)
axes[0].axhline(0, color='red', linestyle='--', linewidth=1.5, label='Break-even (0%)')
axes[0].set_title('Profit Margin Distribution by Cluster')
axes[0].legend()
axes[0].grid(True, linestyle='--', alpha=0.5)

# Right Panel: 2D PCA Scatter plot colored by cluster
sns.scatterplot(x='PCA1', y='PCA2', hue='Cluster', palette='Set1', data=df, ax=axes[1], alpha=0.7)
axes[1].set_title('Cluster Visualization in PCA Space')
axes[1].grid(True, linestyle='--', alpha=0.5)

plt.tight_layout()
plt.show()
```

## Subtask 7: Segment Profiling & Business Action

### 1. Naming & Profiling the Segments
* **Cluster 0: "Stable Standard Shoppers"**  
  *Characteristics:* Moderate purchase volume, low discounts, reliable unit quantities, and consistent margins.  
  *Description:* This segment represents everyday transactions characterized by standard pricing and stable, small-scale positive profits with minimal promotional reliance.

* **Cluster 1: "Bulk Discount Drain"**  
  *Characteristics:* Exceptionally high discount levels, low or negative margins, and elevated unit counts.  
  *Description:* These orders involve bulk quantities of merchandise sold under aggressive promotional rates, resulting in severe profit degradation and negative margins.

* **Cluster 2: "High-Value Premium Buyers"**  
  *Characteristics:* Extremely high average sales volume, superior unit pricing, and robust positive profit margins.  
  *Description:* This represents premium, high-ticket orders where customers acquire high-value items with little to no discount applied, generating the bulk of net revenue.

* **Cluster 3: "Low-Value Opportunists"**  
  *Characteristics:* Minimal unit transaction size, low overall revenue generation, and moderate discount seeking.  
  *Description:* Price-sensitive buyers who make small, sporadic transactions on cheap goods, contributing minimally to both revenue streams and overall margins.

### 2. Strategic Recommendations for the VP of Marketing
* **Action 1 (Risk Mitigation):** "Bulk Discount Drain" (Cluster 1) represents our most critical financial risk due to negative margins driven by aggressive pricing, and we must immediately implement a minimum order volume threshold alongside capping maximum discount limits to protect net profitability.  
* **Action 2 (Growth Strategy):** "High-Value Premium Buyers" (Cluster 2) forms our most profitable segment of full-price buyers; we should launch a dedicated premium loyalty program and offer personalized white-glove support to maximize customer lifetime value and drive repeat purchases.
