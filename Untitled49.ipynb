```python
import pandas as pd
import seaborn as sns

df = sns.load_dataset("tips")

df.head()
```

```python
df.pivot_table(
values='total_bill',
index='sex',
columns='smoker',
aggfunc='mean'
)

```

### Task 1: Business Observations
Based on the pivot table above:
- **Highest Average Bill Segment:** Male smokers have the highest average bill ($22.28).
- **Lowest Average Bill Segment:** Female smokers have the lowest average bill ($17.98).
- **Influence of Smoking Status:** Smoking status appears to influence spending behavior differently based on sex. Male smokers spend more on average than male non-smokers (~$22.28 vs ~$19.79), whereas female smokers spend slightly less on average than female non-smokers (~$17.98 vs ~$18.11).

### Task 2: Cross-Tabulation and Row Percentages

```python
# Part A: Basic frequency table between day and time
ct_freq = pd.crosstab(df['day'], df['time'])
print("--- Part A: Frequency Table ---")
display(ct_freq)
```

```python
# Part B: Row-percentage table using normalize='index'
ct_pct = pd.crosstab(df['day'], df['time'], normalize='index') * 100
print("\n--- Part B: Row-Percentage Table (%) ---")
display(ct_pct)
```

### Task 2: Business Observations
- **Most Dinner Customers:** Saturday is the day with the highest number of dinner customers observed (87 customers), followed closely by Sunday (76 customers).
- **Distribution of Visits:** Customer visits are highly uneven. Lunch is only served on Thursdays and Fridays (with Thursday being almost entirely lunch, and Friday split), while Saturdays and Sundays only have dinner visits.
- **Value of Row-Percentage Table:** While the raw count shows how many customers visit, the row-percentage table highlights the proportion of service within each day. For example, it reveals that Thursday is almost exclusively a lunch day (approx. 98.4% lunch vs 1.6% dinner), and Friday splits more evenly (~22% lunch vs ~78% dinner), which is harder to compare directly using raw frequencies alone.

### Task 3: Manager Recommendation

**Recommendation:**

Pivot data reveals that male smokers represent our highest-spending segment (averaging $22.28 per bill).
