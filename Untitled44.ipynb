```python
# =========================
# Q1 — RUNNING TOTAL
# =========================

q1 = """
WITH monthly AS (
    SELECT
        c.region,
        d.month,
        SUM(f.sale_amount) AS monthly_sales
    FROM FactSales f
    JOIN DimCustomer c
        ON f.customer_id = c.customer_id
    JOIN DimDate d
        ON f.date_id = d.date_id
    GROUP BY c.region, d.month
)
SELECT
    region,
    month,
    monthly_sales,
    SUM(monthly_sales) OVER (
        PARTITION BY region
        ORDER BY month
        ROWS BETWEEN UNBOUNDED PRECEDING AND CURRENT ROW
    ) AS running_total
FROM monthly
ORDER BY region, month;
"""

print("=" * 70)
print("Q1 — RUNNING TOTAL BY REGION AND MONTH")
print("=" * 70)
print(q(q1).to_string(index=False))


# =========================
# Q2 — RANK WITHIN CATEGORY
# =========================

q2 = """
WITH product_revenue AS (
    SELECT
        p.category,
        p.product_id,
        p.product_name,
        SUM(f.sale_amount) AS revenue
    FROM FactSales f
    JOIN DimProduct p
        ON f.product_id = p.product_id
    GROUP BY
        p.category,
        p.product_id,
        p.product_name
),
ranked_products AS (
    SELECT
        category,
        product_id,
        product_name,
        revenue,
        RANK() OVER (
            PARTITION BY category
            ORDER BY revenue DESC
        ) AS category_rank
    FROM product_revenue
)
SELECT
    f.sale_id,
    r.category,
    r.product_name,
    r.revenue,
    r.category_rank
FROM FactSales f
JOIN ranked_products r
    ON f.product_id = r.product_id
ORDER BY r.category, r.category_rank, f.sale_id;
"""

print("\n")
print("=" * 70)
print("Q2 — PRODUCT RANK WITHIN CATEGORY")
print("=" * 70)
print(q(q2).to_string(index=False))


# =========================
# Q3 — MONTH-OVER-MONTH
# =========================

q3 = """
WITH monthly AS (
    SELECT
        c.region,
        d.month,
        SUM(f.sale_amount) AS sales
    FROM FactSales f
    JOIN DimCustomer c
        ON f.customer_id = c.customer_id
    JOIN DimDate d
        ON f.date_id = d.date_id
    GROUP BY c.region, d.month
),
lagged AS (
    SELECT
        region,
        month,
        sales,
        LAG(sales, 1) OVER (
            PARTITION BY region
            ORDER BY month
        ) AS previous_month_sales
    FROM monthly
)
SELECT
    region,
    month,
    sales,
    previous_month_sales,
    sales - previous_month_sales AS month_over_month_change
FROM lagged
ORDER BY region, month;
"""

print("\n")
print("=" * 70)
print("Q3 — MONTH-OVER-MONTH CHANGE")
print("=" * 70)
print(q(q3).to_string(index=False))


# =========================
# Q3 INTERPRETATION
# =========================

print("\n")
print("=" * 70)
print("Q3 — INTERPRETATION")
print("=" * 70)
print(
    "South has the largest positive period-over-period change in the "
    "recorded data, increasing from ₹899 in month 6 to ₹72,000 in month 11, "
    "a change of ₹71,101. This sharp increase is driven by the ₹72,000 "
    "Galaxy S24 sale and indicates a major jump in South-region revenue."
)
```
