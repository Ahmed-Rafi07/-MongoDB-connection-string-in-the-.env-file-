# Live Session Fixes & Assignment: Fare Treatment

This notebook covers the classification, treatment, and final audit of missing values in the Titanic dataset, focusing specifically on the 'Zero Fare' anomaly.

```python
import pandas as pd
import numpy as np

# Load the Titanic dataset
url = 'https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv'
df = pd.read_csv(url)

# Apply typical live session treatments for context
df['Embarked'] = df['Embarked'].fillna(df['Embarked'].mode()[0])
df['Age'] = df['Age'].fillna(df.groupby(['Pclass', 'Sex'])['Age'].transform('median'))
df['Cabin_Assigned'] = df['Cabin'].notna().astype(int)

display(df.head())
```

## Subtask 1: Classify the "Zero Fare" Mechanism

### Classification: **MAR (Missing at Random)**

**Justification:**
The zero-fare tickets are not completely random (MCAR), as they are systematically linked to specific groups. Many of these passengers belong to specific passenger classes (`Pclass`), representing either crew members, promotional/complimentary tickets, or shipyard employees on maiden voyages. Because the probability of having a missing (zero) fare is related to observable group characteristics like ticket class and gender, the mechanism is classified as Missing at Random (MAR).

## Subtask 2: Implement the Treatment for Fare

```python
# Show initial statistics of Fare before treatment
print("--- Fare Summary Statistics (Before Fix) ---")
print(df['Fare'].describe())

# Expose the gaps: replace 0 with NaN
df['Fare'] = df['Fare'].replace(0, np.nan)

# Apply Imputation: MAR treatment using groupby Pclass and filling with median
df['Fare'] = df['Fare'].fillna(df.groupby('Pclass')['Fare'].transform('median'))

# Statistical Check: describe after fix
print("\n--- Fare Summary Statistics (After Fix) ---")
print(df['Fare'].describe())
```

### Observation: Fare Treatment Impact
Replacing the $0$ values with group-conditional medians (based on `Pclass`) shifted the minimum fare from $0 to a realistic baseline and slightly increased the mean fare. This adjustment provides a more honest view of the ship's economy because it eliminates the artificial deflation caused by clerical errors and complimentary tickets, accurately reflecting the true purchasing power and economic distribution of the travelers.

## Subtask 3: Final Investigation Audit

```python
# Check missing values across the entire dataset
print("--- Missing Values Count ---")
print(df.isna().sum())

# Print final dataset shape
print(f"\nFinal Dataset Shape: {df.shape}")
```

### Observation: Final Audit Analysis
* **Columns with 0 missing values:** `PassengerId`, `Survived`, `Pclass`, `Name`, `Sex`, `Age`, `SibSp`, `Parch`, `Ticket`, `Fare`, `Embarked`, and our new engineered column `Cabin_Assigned` all show 0 missing values.
* **Cabin Column Analysis:** The original `Cabin` column still contains missing values. This is completely acceptable because the missingness rate is extremely high (~77%), making any direct imputation highly unreliable and prone to introducing noise. Instead of utilizing the incomplete raw `Cabin` column, analysts should use the engineered `Cabin_Assigned` column (which binary-encodes whether a passenger had an assigned cabin) for downstream modeling tasks.
